1.  What are Type I and Type II errors in hypothesis testing, and how do they impact decision-making?
ans Type I and Type II errors are two fundamental concepts in hypothesis testing that describe the kinds of mistakes that can occur when drawing conclusions from statistical data. These errors directly influence decision-making and impact how confident one can be in research findings.
Type I Error Explained
A Type I error, also known as a "false positive," happens when the null hypothesis is true, but is incorrectly rejected. This means the test indicates that an effect or difference exists when, in reality, it does not. The probability of committing a Type I error is denoted by
α
α (alpha), commonly set to 0.05 (5%) in many experiments.​

Example: In a clinical trial, concluding a drug works when it actually does not constitutes a Type I error.​

Impact: Type I errors can lead to unnecessary changes or new treatments being adopted based on incorrect evidence.​
Type II Error Explained
A Type II error, or "false negative," occurs when the null hypothesis is false but is incorrectly not rejected. This means the test fails to detect a real effect or difference. The probability for this error is represented by
β
β (beta).​

Example: In medical testing, failing to identify that a drug is effective when it actually is is a Type II error.​

Impact: Type II errors can lead to missed opportunities, such as not adopting a useful innovation or drug because the evidence did not show a difference, even though one existed.

2. What is the P-value in hypothesis testing, and how should it be interpreted in the context of the null hypothesis?
ans The P-value in hypothesis testing is a measure of the probability that observed results (or more extreme results) would occur if the null hypothesis were true. It quantifies the strength of evidence against the null hypothesis by indicating how likely the observed data would be under the assumption that there is no real effect or difference.​

Definition and Calculation
The P-value is calculated from statistical test results and represents the likelihood of obtaining results as extreme or more extreme than those observed, under the null hypothesis.​

A smaller P-value means there is stronger evidence against the null hypothesis.

3. Explain the difference between a Z-test and a T-test, including when to use each.
ans Z-test
Use a Z-test when the sample size is large (n ≥ 30) and the population variance (or standard deviation) is known.​

It relies on the assumption that the sampling distribution of the mean is approximately normal.

The Z-test uses the standard normal distribution (Z-distribution) for calculations.

Common applications: quality control, large-scale clinical trials, and cases where population parameters are established.
T-test
Use a T-test when the sample size is small (n < 30) and/or the population variance is unknown.​

The T-test uses the sample standard deviation as an estimate of the population standard deviation.

It relies on the t-distribution, which is broader and changes depending on sample size (degrees of freedom), accounting for more variability in smaller samples.

Common applications: small experimental studies, pilot data, or situations where population variance is not available.

ans A confidence interval is a statistical range, usually defined by an upper and lower bound, that is likely to contain the true value of a population parameter (such as the mean) with a specified degree of certainty known as the confidence level—commonly 95% or 99%. It provides a measure of estimate precision and is centered around a sample statistic (e.g., the sample mean).

5. Describe the purpose and assumptions of an ANOVA test. How does it extend hypothesis testing to more than two groups?
ans An ANOVA (Analysis of Variance) test allows researchers to determine whether there are statistically significant differences between the means of three or more groups, extending hypothesis testing beyond just two groups.​

Purpose of ANOVA
The primary purpose of ANOVA is to assess if at least one group mean is different from the others by comparing the variance between groups to variance within groups.​

If the test finds significant differences overall, post-hoc tests are often used to pinpoint which groups differ.​
Key Assumptions
ANOVA relies on several assumptions to ensure its results are statistically valid:​

Normality: The data within each group should be normally distributed.

Homogeneity of variance: Group variances should be approximately equal.


```python
#Write a Python program to perform a one-sample Z-test and interpret the result for a given dataset.
import numpy as np
from scipy.stats import norm

def one_sample_z_test(sample_data, population_mean, population_std, alpha=0.05):
    n = len(sample_data)
    sample_mean = np.mean(sample_data)
    # Calculate standard error
    se = population_std / np.sqrt(n)
    # Calculate Z-statistic
    z = (sample_mean - population_mean) / se
    # Calculate two-tailed p-value
    p_value = 2 * (1 - norm.cdf(abs(z)))
    # Interpret result
    if p_value < alpha:
        interpretation = "Reject the null hypothesis: The sample mean is significantly different from the population mean."
    else:
        interpretation = "Fail to reject the null hypothesis: No significant difference between sample and population mean."
    return {
        'sample_mean': sample_mean,
        'population_mean': population_mean,
        'population_std': population_std,
        'z_statistic': z,
        'p_value': p_value,
        'alpha': alpha,
        'interpretation': interpretation
    }

# Example: Run the test on a sample dataset
sample_data = [54, 56, 60, 62, 53, 59, 57, 61, 58, 55]
population_mean = 55
population_std = 4
result = one_sample_z_test(sample_data, population_mean, population_std)
print(result)

```
